**Escribimos un fichero HQL para crear nuestra tabla. Como los datos que tenemos tienen encabezados usamos la cláusula TBLPROPERTIES para ignorar la primera fila.**

```python
%%writefile crear-tabla-stars.hql
CREATE TABLE stars(
                 Name STRING,
                 Constellation STRING,
                 BayernDesignation STRING,
                 Designation STRING,
                 ApprovalDate DATE
            )
            ROW FORMAT DELIMITED
            FIELDS TERMINATED BY ','
            STORED AS TEXTFILE
            LOCATION '/ejemplo-hive/'
            TBLPROPERTIES ('skip.header.line.count'='1');
```

**Ejecutamos el fichero HQL para crear la tabla**

```python
# ejecucion del fichero hql mediante beeline
!beeline -u "jdbc:hive2://" -f crear-tabla-stars.hql
```

**Mostramos las tablas que hay**

```python
!beeline -u "jdbc:hive2://" -e "show tables;"
```

**Ahora cargaremos los datos en la tabla. Ya vimos como se puede cargar los datos en el momento de crear la tabla, en este caso lo hacemos posteriormente para ver otras opciones de carga.**

```python
#creamos la carpeta
! hdfs dfs -mkdir /ejemplo-hive

#subimos el fichero stars_names.csv
! hdfs dfs -put /media/notebooks/ejemplo-hive/Stars_Names.csv /ejemplo-hive

#vemos que se haya cargado correctamente
! hdfs dfs -ls /ejemplo-hive
```

```python
#cargamos los datos dentro de la tabla
! beeline -u "jdbc:hive2://" -e \
"LOAD DATA INPATH '/ejemplo-hive/Stars_Names.csv' \
INTO TABLE stars;"
```

**Mediante una consulta sencilla visualizamos que los datos se han cargado correctamente.**

```python
!beeline -u "jdbc:hive2://" -e \
"SELECT * FROM stars LIMIT 3;"
```

**Realizamos las consultas para comprobar el funcionamiento de Hive.**

```python
!beeline -u "jdbc:hive2://" -e \
"SELECT * FROM stars \
WHERE YEAR(approvaldate) = 2016 and MONTH(approvaldate) = 11;"
```

**Escribimos una consulta en un fichero HQL para ver el funcionamiento también de esta forma.**

```python
%%writefile consulta.hql
SELECT COUNT(*) as num_estrellas, constellation
FROM stars
GROUP BY constellation
ORDER BY num_estrellas DESC
LIMIT 5;
```

```python
!beeline -u "jdbc:hive2://" -f consulta.hql
```
