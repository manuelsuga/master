# <font color="#004D7F"> 1. Algoritmo Selección de miembros del conjunto</font>

- El problema con un conjunto de promedio ponderado es que todos los modelos siguen siendo parte del conjunto (sin importar si son malos clasificadores).
- El enfoque general de elegir u optimizar automáticamente los miembros de un conjunto se denomina __selección de conjunto__. Dos enfoques comunes incluyen el conjunto creciente y conjunto poda.
- Da modelos menos complejos (menor tamaño) y/o mejor rendimiento predictivo
- A veces se prefiere una tener una pequeña caída en la métrica si reducimos considerablemente la complejidad.

#### Conjunto creciente (_Ensemble Growing_) 
- Agrega miembros al conjunto hasta que no se observen más mejoras.
- El modelo comienza sin miembros e implica agregar nuevos miembros hasta que no se observan más mejoras.
- Esto podría realizarse de una manera codiciosa donde los miembros se agregan uno a la vez solo si dan como resultado una mejora en el rendimiento del modelo.
- Es más eficiente en los casos en los que se espera que un gran número de miembros del conjunto se desempeñen mejor.
    
#### Conjunto poda (_Ensemble Pruning_):
- Retire los miembros del conjunto hasta que no se observen más mejoras.
- El modelo comienza con todos los miembros posibles que se están considerando y elimina miembros del conjunto hasta que no se observan más mejoras.
- Esto podría realizarse de una manera codiciosa, donde los miembros son eliminados uno a la vez y sólo si su eliminación da como resultado un aumento en el desempeño del conjunto en general.
-Es más eficiente en los casos en los que se espera que un pequeño número de miembros del conjunto se desempeñen mejor.
