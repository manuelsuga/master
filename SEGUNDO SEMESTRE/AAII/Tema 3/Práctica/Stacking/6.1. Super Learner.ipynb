# <font color="#004D7F">0. Contexto</font>

Super Learner es una aplicación de Stacking a la validación cruzada de k veces, donde todos los modelos usan las mismas divisiones de datos de k veces y se ajusta un metamodelo a las predicciones de cada uno. En este tutorial, aprenderemos:
- Super Learner es la aplicación de Stacking utilizando predicciones fuera de fold durante la validación cruzada de k veces.
- Super Learner es sencillo de implementar en Python utilizando modelos scikit-learn.
- La biblioteca ML-Ensemble proporciona una implementación conveniente que permite que Super Learner se adapte y utilice con solo unas pocas líneas de código.

# <font color="#004D7F"> 1. Algoritmo Super Learner</font>

- Super Learner implica primero predefinir las _k_ divisiones de sus datos y luego evaluar todos los diferentes algoritmos y configuraciones en las mismas divisiones de los datos. 
- Luego, todas las predicciones se guardan y se utilizan para entrenar un modelo que aprende cómo combinar mejor las predicciones.
- El objetivo es que Super Learner mejore cualquier modelo contribuyente.
- Fué propuesto por Mark van der Laan _et al._ en su artículo llamado _"Super Learner"_. 

El procedimiento de Super Learner se resumen en los siguientes pasos (ver Figura):
1. Seleccionar una división de _k_ veces del conjunto de datos de entrenamiento.
2. Seleccione _m_ modelos base o configuraciones de modelo.
3. Para cada modelo base:
  - Evalúe utilizando validación cruzada de _k_ veces.
  - Almacene todas las predicciones desplegadas.
  - Ajuste el modelo en el conjunto de datos de entrenamiento completo y guárdelo.
4. Ajuste un metamodelo a las predicciones desplegadas.
5. Evalúe el modelo en un conjunto de datos reservado o utilice el modelo para hacer predicciones.

- La función `get_out_of_fold_predictions()` predice valores numérica meidante la función `predict()`


<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> __Nota__: Más información sobre [`hstack`](https://numpy.org/doc/stable/reference/generated/numpy.hstack.html)

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> __Nota__: Más información sobre [`vstack`](https://numpy.org/doc/stable/reference/generated/numpy.vstack.html)

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> __Nota__: Más información sobre [`asarray`](https://numpy.org/doc/stable/reference/generated/numpy.asarray.html)

```python
from numpy import hstack
from numpy import vstack
from numpy import asarray
from sklearn.model_selection import KFold

def get_out_of_fold_predictions(X, y, models):
	meta_X, meta_y = list(), list()
	kfold = KFold(n_splits=10, shuffle=True)
	for train_ix, test_ix in kfold.split(X):
		fold_yhats = list()
		train_X, test_X = X[train_ix], X[test_ix]
		train_y, test_y = y[train_ix], y[test_ix]
		meta_y.extend(test_y)
		for model in models:
			model.fit(train_X, train_y)
			yhat = model.predict(test_X)
			fold_yhats.append(yhat.reshape(len(yhat),1))
		meta_X.append(hstack(fold_yhats))
	return vstack(meta_X), asarray(meta_y)
```

Ajustamos todos los modelos base en todo el conjunto de datos de entrenamiento.

```python
# fit all base models on the training dataset
def fit_base_models(X, y, models):
	for model in models:
		model.fit(X, y)
```

Ajustamos el metamodelo al dataset preparado. En este caso, utilizaremos LoR como metamodelo.

<div class="alert alert-block alert-info">
    
<i class="fa fa-info-circle" aria-hidden="true"></i> __Nota__: Más información sobre la clase [`LinearRegression`](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html)
</div>

```python
from sklearn.linear_model import LinearRegression

def fit_meta_model(X, y):
	model = LinearRegression()
	model.fit(X, y)
	return model
```

Evaluamos los modelos base en el conjunto de datos reservados e informar el error absoluto medio (MAE).

```python
from sklearn.metrics import mean_absolute_error

def evaluate_models(X, y, models):
	for model in models:
		yhat = model.predict(X)
		mae = mean_absolute_error(y, yhat)
		print('%s: MAE %.3f' % (model.__class__.__name__, mae))
```

Utilizamos Super Lerarner (base y metamodelo) para hacer predicciones sobre el conjunto de datos reservados y evaluar el rendimiento. 

```python
# make predictions with stacked model
def super_learner_predictions(X, models, meta_model):
	meta_X = list()
	for model in models:
		yhat = model.predict(X)
		meta_X.append(yhat.reshape(len(yhat),1))
	meta_X = hstack(meta_X)
	return meta_model.predict(meta_X)
```

Llamamos a las funciones para obtener los modelos y preparar el dataset del metamodelo.

```python
X, y = get_dataset()
X, X_val, y, y_val = train_test_split(X, y, test_size=0.50)
print('Train', X.shape, y.shape, 'Test', X_val.shape, y_val.shape)
models = get_models()
meta_X, meta_y = get_out_of_fold_predictions(X, y, models)
print('Meta ', meta_X.shape, meta_y.shape)
fit_base_models(X, y, models)
meta_model = fit_meta_model(meta_X, meta_y)
evaluate_models(X_val, y_val, models)
yhat = super_learner_predictions(X_val, models, meta_model)
score = mean_absolute_error(y_val, yhat)
print('Super Learner: MAE %.3f' % score)
```

Esto significa que los errores positivos de gran magnitud se convierten en grandes errores negativos (por ejemplo, 100 se convierte en -100) y un modelo perfecto no tiene ningún error con un valor de 0,0.
